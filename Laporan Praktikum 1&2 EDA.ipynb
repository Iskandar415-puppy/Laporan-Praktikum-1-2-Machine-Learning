```python
# Iskandar Zulkarnain Wijaya_202331121
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Iskandar Zulkarnain Wijaya_202331121
df = pd.read_csv("Expanded_Dataset.csv")
print(df.head())
```

```python
# Iskandar Zulkarnain Wijaya_202331121
df.info()
```

```python
# Iskandar Zulkarnain Wijaya_202331121
print(df.describe())
```

```python
# Iskandar Zulkarnain Wijaya_202331121
print(df.isna().sum())
```

```python
# Iskandar Zulkarnain Wijaya_202331121
print("Quarter:", df["Quarter"].unique())
print("Product Model:", df["Product Model"].unique())
print("5G Capability:", df["5G Capability"].unique())
print("Region:", df["Region"].unique())
```

```python
# Iskandar Zulkarnain Wijaya_202331121
# Mengisi kolom numerik dengan mean, dan kolom kategorik dengan modus
for col in df.select_dtypes(include=["float64", "int64"]).columns:
  df[col] = df[col].fillna(df[col].mean())

for col in df.select_dtypes(include=["object"]).columns:
  df[col] = df[col].fillna(df[col].mode()[0])

print("Data kosong setelah diisi:")
print(df.isna().sum().sum())
```

```python
# Iskandar Zulkarnain Wijaya_202331121
print("Jumlah data duplikat awal:", df.duplicated().sum())
df = df.drop_duplicates()
df.info()
```

```python
# Iskandar Zulkarnain Wijaya_202331121
plt.figure(figsize=(10, 5))
df["Region"].value_counts().plot(kind="bar", color="skyblue", edgecolor="black")
plt.title("Jumlah Data Berdasarkan Region", fontsize=14)
plt.xlabel("Region", fontsize=12)
plt.ylabel("Jumlah", fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```
